### C2

### C3

w_hh = 0.5, w_xh = 1.0, h_0 = 0 이고 입력이 x_1=1, x_2=0, x_3=0 일 때 h_1, h_2, h_3 을 구하시오.
(h_t = tanh(w_hh·h_{t-1} + w_xh·x_t), tanh(1)≈0.7616 사용)

**정답: h1≈0.7616, h2≈0.3640, h3≈0.1799**

```
h_1 = tanh(0.5·0 + 1·1) = tanh(1) ≈ 0.7616
h_2 = tanh(0.5·0.7616 + 0) = tanh(0.3808) ≈ 0.3640
h_3 = tanh(0.5·0.3640 + 0) = tanh(0.1820) ≈ 0.1799
→ 첫 입력의 흔적이 매 스텝 절반 이하로 줄어든다. 이것이 기울기 소실의 직관적 모습이다.
```

### C4

|w_hh| = 0.9 일 때, 20스텝 떨어진 과거로 전달되는 기울기 배율 0.9^20 은 대략 얼마인가? 그리고 |w_hh| = 1.1 이면 1.1^20 은 대략 얼마인가?

**정답: 0.9^20≈0.12, 1.1^20≈6.7**

```
0.9^20 ≈ 0.1216 (약 8분의 1로 감소 → 기울기 소실)
1.1^20 ≈ 6.7275 (약 7배로 증가 → 기울기 폭발)
1에서 0.1만 벗어나도 20스텝 만에 이만큼 벌어진다. 문장이 100단어면 0.9^100 ≈ 2.7e-5 로 완전히 사라진다.
```

### C5

LSTM에서 f_t = 0.9, C_{t-1} = 2.0, i_t = 0.5, C̃_t = 1.0 일 때 C_t 를 구하시오.
이어서 o_t = 0.8, tanh(C_t) ≈ 0.9866 일 때 h_t 를 구하시오.

**정답: C_t = 2.3, h_t ≈ 0.789**

```
C_t = f_t·C_{t-1} + i_t·C̃_t = 0.9·2.0 + 0.5·1.0 = 1.8 + 0.5 = 2.3
h_t = o_t · tanh(C_t) = 0.8 · 0.9866 ≈ 0.7893
→ cell state에는 2.3이 저장되지만, 밖으로 나가는 것은 output gate가 걸러낸 0.79뿐이다.
```

### C6

forget gate가 매 스텝 0.8로 일정할 때, 첫 스텝에 기록한 기억이 절반 이하로 줄어드는 것은 몇 스텝 뒤인가?
(log 0.5 / log 0.8 ≈ 3.106)

**정답: 4스텝**

```
0.8^t < 0.5 를 만족하는 최소 정수 t 를 찾는다. t > log0.5/log0.8 ≈ 3.106 이므로 t = 4.
0.8^3 = 0.512 (아직 절반 이상), 0.8^4 = 0.4096 (절반 미만).
→ forget gate 값이 곧 '기억의 반감기'를 정한다.
```

```

→ A가 선택된다. 첫 단어만 보면 B(0.6)가 더 높지만, 전체를 보면 A가 낫다. 이것이 Greedy가 실패하고 Beam Search가 필요한 이유다.
```

### C12

## 파트 D. 코드 구현 — 모범 답안

````python
def one_hot(word, vocab):
    v = np.zeros(len(vocab))
    v[vocab.index(word)] = 1.0
    return v


def skipgram_pairs(tokens, center, window):
    lo = max(0, center - window)
    hi = min(len(tokens) - 1, center + window)
    return [(tokens[center], tokens[i]) for i in range(lo, hi + 1) if i != center]


def rnn_forward(xs, w_hh, w_xh):
    h, out = 0.0, []
    for x in xs:
        h = np.tanh(w_hh * h + w_xh * x)     # recurrence: 같은 가중치를 매 시점 재사용
        out.append(h)
    return out


def lstm_step(x, h, C, p):
    z = np.concatenate([h, x])                   # [h_{t-1}, x_t]
    f = sigmoid(p['Wf'] @ z + p['bf'])           # forget gate: 무엇을 버릴까
    i = sigmoid(p['Wi'] @ z + p['bi'])           # input gate : 얼마나 쓸까
    g = np.tanh(p['Wc'] @ z + p['bc'])           # new cell content
    C_new = f * C + i * g                        # 곱셈 + 덧셈 = 기울기 지름길
    o = sigmoid(p['Wo'] @ z + p['bo'])           # output gate: 얼마나 내보낼까
    return o * np.tanh(C_new), C_new


def softmax(x):
    x = np.asarray(x, dtype=float)
    finite = x[np.isfinite(x)]
    if finite.size:
        x = x - finite.max()                     # 오버플로 방지
    e = np.where(np.isfinite(x), np.exp(x), 0.0) # -inf 는 정확히 0
    return e / e.sum()


def attention(s, H):
    e = H @ s                                    # (1) score
    a = softmax(e)                               # (2) 분포
    return a, a @ H                              # (3) 가중합


def scaled_dot_attention(Q, K, V, mask=False):
    d = Q.shape[-1]
    scores = Q @ K.T / np.sqrt(d)                # sqrt(d) 스케일링
    if mask:
        m = np.triu(np.ones_like(scores), k=1).astype(bool)
        scores = np.where(m, -np.inf, scores)    # 미래 차단
    scores = scores - scores.max(axis=-1, keepdims=True)
    A = np.exp(scores)
    A = A / A.sum(axis=-1, keepdims=True)
    return A @ V, A


def positional_encoding(max_len, d):
    pos = np.arange(max_len)[:, None]
    i = np.arange(d)[None, :]
    angle = pos / np.power(10000.0, (2 * (i // 2)) / d)
    return np.where(i % 2 == 0, np.sin(angle), np.cos(angle))


def ngram_prob(tokens, n, context, word):
    ctx = tuple(context[-(n - 1):]) if n > 1 else ()
    full = ctx + (word,)
    if ctx:
        c_ctx = sum(1 for i in range(len(tokens) - len(ctx) + 1)
                    if tuple(tokens[i:i + len(ctx)]) == ctx)
    else:
        c_ctx = len(tokens)
    c_full = sum(1 for i in range(len(tokens) - len(full) + 1)
                 if tuple(tokens[i:i + len(full)]) == full)
    return c_full / c_ctx if c_ctx else 0.0      # 분모 0 -> 확률 계산 불가 (희소성)


def bert_mask_decide(rng, p=0.15, mask_ratio=0.8):
    if rng.random() >= p:
        return 'keep'                            # 85%: 선택 안 됨
    r = rng.random()
    if r < mask_ratio:
        return 'mask'                            # 15% x 80% = 12%
    if r < mask_ratio + (1 - mask_ratio) / 2:
        return 'random'                          # 15% x 10% = 1.5%
    return 'same'                                # 15% x 10% = 1.5%

print("모범 답안 10개 정의 완료")
````

### 파트 D 전체 검사 — 문제집의 assert를 그대로 실행

````python
# ----- D1 -----
v = ['나','는','밥','을','먹었다']
assert one_hot('밥', v).tolist() == [0,0,1,0,0]
assert one_hot('나', v).tolist() == [1,0,0,0,0]
assert float(one_hot('밥', v) @ one_hot('나', v)) == 0.0
assert one_hot('먹었다', v).sum() == 1
print("D1 통과 — 서로 다른 단어의 내적이 0인 것도 확인했습니다.")

# ----- D2 -----
t = '어제 카페 갔었어 거기 사람 많더라'.split()
assert skipgram_pairs(t, 2, 2) == [('갔었어','어제'),('갔었어','카페'),('갔었어','거기'),('갔었어','사람')]
assert len(skipgram_pairs(t, 0, 2)) == 2      # 문장 맨 앞은 왼쪽 문맥이 없음
assert len(skipgram_pairs(t, 5, 2)) == 2      # 맨 뒤도 마찬가지
assert len(skipgram_pairs(t, 3, 5)) == 5      # 윈도우가 문장보다 크면 전부
print("D2 통과 — 문장 경계에서 윈도우가 잘리는 것도 처리했습니다.")

# ----- D3 -----
out = rnn_forward([1,0,0], 0.5, 1.0)
assert len(out) == 3
assert abs(out[0] - np.tanh(1.0)) < 1e-9
assert abs(out[1] - np.tanh(0.5*np.tanh(1.0))) < 1e-9
assert abs(out[2] - np.tanh(0.5*out[1])) < 1e-9
assert abs(rnn_forward([1,0,0], 0.0, 1.0)[1]) < 1e-12   # w_hh=0 이면 과거를 못 기억함
print("D3 통과 — w_hh=0 이면 기억이 사라지는 것도 확인했습니다.")

# ----- D4 -----
rng = np.random.default_rng(0)
dh, dx = 3, 2
p = {}
for k in ['f','i','c','o']:
    p['W'+k] = rng.normal(0, 0.5, (dh, dh+dx))
    p['b'+k] = np.zeros(dh)
h0, C0, x0 = np.zeros(dh), np.zeros(dh), rng.normal(0,1,dx)
h1, C1 = lstm_step(x0, h0, C0, p)
assert h1.shape == (dh,) and C1.shape == (dh,)

# forget gate를 강제로 열고 input gate를 닫으면 cell state가 그대로 유지되어야 함
p2 = {k: v.copy() for k, v in p.items()}
p2['Wf'][:] = 0; p2['bf'][:] =  50      # f ~ 1
p2['Wi'][:] = 0; p2['bi'][:] = -50      # i ~ 0
C_start = np.array([1.0, -2.0, 0.5])
_, C_ = lstm_step(x0, h0, C_start, p2)
assert np.allclose(C_, C_start, atol=1e-6), "f~1, i~0 이면 C_t = C_{t-1} 이어야 합니다"

# 반대로 forget을 닫으면 과거가 지워져야 함
p3 = {k: v.copy() for k, v in p.items()}
p3['Wf'][:] = 0; p3['bf'][:] = -50
p3['Wi'][:] = 0; p3['bi'][:] = -50
_, C_ = lstm_step(x0, h0, C_start, p3)
assert np.allclose(C_, 0, atol=1e-6), "f~0, i~0 이면 과거가 지워져야 합니다"
print("D4 통과 — 게이트를 극단으로 두면 기억이 유지/삭제되는 것까지 확인했습니다.")

# ----- D5 -----
assert abs(softmax([1,2,3]).sum() - 1.0) < 1e-12
assert np.allclose(softmax([0,0,0]), [1/3,1/3,1/3])
r = softmax([1000, 1001, 1002])                 # 오버플로가 나면 안 됨
assert np.all(np.isfinite(r)) and abs(r.sum()-1) < 1e-12
r2 = softmax([1.0, -np.inf, 2.0])               # -inf 는 정확히 0 (masking에 쓰임)
assert r2[1] == 0.0
print("D5 통과 — 오버플로 방지와 -inf 처리까지 확인했습니다.")

# ----- D6 -----
H = np.array([[1.,0.],[0.,1.],[1.,1.]])
s = np.array([0.,3.])
a, ctx = attention(s, H)
assert abs(a.sum() - 1.0) < 1e-12
assert int(np.argmax(a)) == 1, "s와 방향이 가장 비슷한 h2가 가장 큰 가중치를 받아야 합니다"
assert np.allclose(ctx, a @ H)
a2, _ = attention(np.array([1.,1.]), np.ones((4,2)))   # 모두 같으면 균등
assert np.allclose(a2, 0.25)
print("D6 통과")

# ----- D7 -----
rng = np.random.default_rng(1)
n, d = 5, 4
Q, K, V = rng.normal(0,1,(n,d)), rng.normal(0,1,(n,d)), rng.normal(0,1,(n,d))
out, A = scaled_dot_attention(Q, K, V)
assert out.shape == (n, d) and A.shape == (n, n)
assert np.allclose(A.sum(axis=1), 1.0)

out_m, A_m = scaled_dot_attention(Q, K, V, mask=True)
up = np.triu(np.ones((n,n)), k=1).astype(bool)
assert np.allclose(A_m[up], 0.0), "미래 위치의 가중치는 정확히 0이어야 합니다"
assert np.allclose(A_m.sum(axis=1), 1.0), "마스킹 후에도 각 행의 합은 1이어야 합니다"
assert abs(A_m[0,0] - 1.0) < 1e-12, "첫 토큰은 자기 자신만 볼 수 있으므로 1.0"

_, A2 = scaled_dot_attention(Q*np.sqrt(d), K, V)      # 스케일링을 실제로 했는지
assert not np.allclose(A, A2), "sqrt(d) 로 나누는 스케일링이 적용되어야 합니다"
print("D7 통과 — 마스킹 후 행 합이 1로 재분배되는 것까지 확인했습니다.")

# ----- D8 -----
PE = positional_encoding(50, 16)
assert PE.shape == (50, 16)
assert np.all(np.abs(PE) <= 1.0 + 1e-12)
assert abs(PE[0,0] - 0.0) < 1e-12 and abs(PE[0,1] - 1.0) < 1e-12   # pos=0 -> sin0=0, cos0=1
assert not np.allclose(PE[3], PE[7])                                # 위치마다 달라야 함
def _c(a,b): return float(a@b/(np.linalg.norm(a)*np.linalg.norm(b)))
assert _c(PE[10], PE[11]) > _c(PE[10], PE[40])                      # 가까울수록 비슷
print("D8 통과 — 가까운 위치일수록 유사도가 높은 성질까지 확인했습니다.")

# ----- D9 -----
toks = ('a b c a b c a b d ' * 3).split()
p_c = ngram_prob(toks, 3, ['a','b'], 'c')
p_d = ngram_prob(toks, 3, ['a','b'], 'd')
assert abs(p_c - 6/9) < 1e-9, f"기대 0.667, 실제 {p_c}"
assert abs(p_d - 3/9) < 1e-9, f"기대 0.333, 실제 {p_d}"
assert abs(p_c + p_d - 1.0) < 1e-9, "확률의 합은 1이어야 합니다"
assert ngram_prob(toks, 3, ['z','z'], 'a') == 0.0     # 본 적 없는 문맥
print("D9 통과 — 본 적 없는 문맥에서 0을 반환하는 것(희소성 문제)도 확인했습니다.")

# ----- D10 -----
from collections import Counter
rng = np.random.default_rng(0)
c = Counter(bert_mask_decide(rng) for _ in range(200000))
tot = sum(c.values())
assert abs(c['keep']/tot   - 0.85 ) < 0.01, f"선택 안 됨 85% 기대, 실제 {c['keep']/tot:.3f}"
assert abs(c['mask']/tot   - 0.12 ) < 0.01, f"[MASK] 12% 기대, 실제 {c['mask']/tot:.3f}"
assert abs(c['random']/tot - 0.015) < 0.005
assert abs(c['same']/tot   - 0.015) < 0.005
print("D10 통과 — 전체 기준 12% / 1.5% / 1.5% 비율이 나왔습니다.")
print("  ", {k: round(v/tot, 4) for k, v in c.items()})

print('\n>>> D1~D10 모든 검사 통과')
````

### E1. 원-핫 인코딩의 두 가지 문제점을 쓰고, 워드 임베딩이 각각을 어떻게 해결하는지 설명하시오.

> ① 차원의 저주 — 어휘 크기만큼의 고차원 희소 벡터라 메모리를 많이 쓰고 데이터가 성기다. 워드 임베딩은 100~300차원의 밀집 벡터로 압축한다.
>
> ② 의미적 정보 부족 — 모든 단어 쌍이 직교해 유사도를 잴 수 없다. 워드 임베딩은 비슷한 문맥에 쓰이는 단어를 공간상 가깝게 배치해 의미적 유사성을 반영한다.

### E2. RNN이 모든 시점에서 같은 가중치를 공유하는 이유를 '파라미터 효율성' 외의 관점에서 설명하시오.

> 가장 중요한 이유는 가변 길이 처리다. 시점마다 다른 W를 쓰면 최대 길이만큼 가중치를 미리 준비해야 하고, 학습 때 본 적 없는 더 긴 입력은 처리할 수 없다. 또 'k번째 자리를 읽는 방법'은 k번째에 온 데이터로만 학습되어 데이터가 극단적으로 부족해진다. 공유하면 '한 단어를 읽어 기억을 갱신하는 방법' 하나만 배워 길이에 상관없이 재사용할 수 있다(CNN이 필터를 공유하는 것과 같은 논리).

### E3. LSTM이 RNN의 기울기 소실을 완화하는 원리를 cell state 수식을 근거로 설명하시오.

> RNN은 h_t = tanh(W h_{t-1} + ...) 이라 역전파에서 W와 tanh 미분(<1)이 계속 곱해져 지수적으로 0에 수렴한다. 반면 LSTM은 C_t = f_t·C_{t-1} + i_t·C̃_t 로, cell state가 곱셈과 덧셈만 거쳐 지나가는 경로를 갖는다. 이 경로에서 ∂C_t/∂C_{t-1} = f_t 이므로 tanh 미분이 끼어들지 않고, forget gate가 1에 가까우면 기울기가 거의 그대로 전달된다. 즉 '곱셈의 사슬'에 '덧셈 지름길'을 낸 것이다.

### E4. N-gram 언어모델에서 n을 크게 하지 못하는 이유를 희소성 관점에서 설명하시오.

> 어휘 크기 V일 때 가능한 n-gram 조합은 V^n 으로 지수적으로 늘어난다. V=5만이면 4-gram만 해도 6×10^18 가지인데, 어떤 코퍼스도 이를 채울 수 없다. 그래서 n이 커질수록 count가 0인 문맥이 폭증해 확률 계산 자체가 불가능해지고(분모 0), 저장 공간도 감당할 수 없다. 즉 '긴 문맥을 보고 싶다'와 '통계를 셀 수 있다'가 충돌한다. 신경망 언어모델은 세는 대신 벡터로 일반화해 이 충돌을 피한다.

### E5. Teacher Forcing의 장점과, 그로 인해 생기는 문제(exposure bias)를 설명하시오.

> 장점: 학습 초반 모델의 예측이 부정확할 때 디코더 입력으로 정답을 넣어주면 오류가 누적되지 않아 학습이 안정적이고 빠르다.
>
> 문제: 학습 때는 항상 정답 뒤를 이어 쓰는 연습만 하는데, 추론 때는 자기가 만든(틀릴 수도 있는) 단어 뒤를 이어 써야 한다. 즉 학습 분포와 추론 분포가 다르다. 한 번 틀리면 모델이 겪어본 적 없는 입력이 들어와 오류가 눈덩이처럼 커진다. 완화책으로 scheduled sampling, beam search 등이 있다.

### E6. Seq2Seq의 bottleneck problem이 무엇이고 Attention이 이를 어떻게 해결하는지 설명하시오.

> 인코더가 입력 문장 전체를 마지막 hidden state 하나(고정 길이 벡터)에 압축하기 때문에, 문장이 길수록 정보 손실이 커진다. 이것이 bottleneck problem이며, 실제로 attention 없는 모델은 문장 길이가 30을 넘으면 BLEU가 급격히 떨어진다.
>
> Attention은 디코더가 매 시점 인코더의 모든 hidden state에 직접 접근해, 지금 필요한 부분에 가중치를 두고 가중합(context vector)을 만든다. 정보가 한 벡터를 통과할 필요가 없어져 병목이 사라진다. 덤으로 멀리 떨어진 단어를 직접 연결하므로 기울기 소실도 완화된다.

### E7. RNN의 두 가지 한계를 쓰고, Self-Attention이 각각을 어떻게 해결하는지 설명하시오.

> ① 장기 의존성 — 먼 단어 쌍이 상호작용하려면 시퀀스 길이만큼의 단계를 거쳐야 해(O(n)) 기울기가 소실된다. Self-Attention은 모든 단어가 각 층에서 직접 연결되므로 최대 상호작용 거리가 O(1)이다.
>
> ② 병렬화 불가 — 앞 시점 계산이 끝나야 다음을 계산할 수 있어 GPU를 못 쓴다. Self-Attention은 모든 위치의 계산이 서로 독립이라 행렬 연산 한 번으로 병렬 처리된다. 이 덕분에 대규모 사전학습이 가능해졌다.

### E8. Self-Attention의 세 가지 한계와 각각의 해결책을 설명하시오.

> ① 순서 정보 부재 — 유사도만 계산하므로 단어 순서를 모른다(순열 등변). → Positional Encoding을 단어 임베딩에 더해 위치 정보를 주입한다.
>
> ② 비선형성 부족 — attention은 가중 평균, 즉 선형 결합이라 복잡한 패턴을 담기 어렵다. → 각 위치에 Feed-Forward Network(FC + ReLU)를 추가한다.
>
> ③ 미래 참조 — 생성 시 아직 만들지 않은 단어를 본다. → Masked Self-Attention으로 미래 위치의 점수를 -∞로 만들어 softmax 후 0이 되게 한다.

### E9. Scaled Dot Product에서 √(d/h) 로 나누는 이유를 설명하시오.

> Query와 Key의 차원이 커질수록 두 벡터의 내적 값과 그 분산도 함께 커진다. 점수 값이 크면 softmax 출력이 거의 one-hot에 가까울 만큼 뾰족해지고, 그 지점에서 softmax의 기울기가 0에 가까워져 학습이 멈춘다(예: softmax([40,50,45]) ≈ [0.000045, 0.993, 0.0067]). √(d/h)로 나누면 점수의 분산이 1 부근으로 정규화되어 분포가 완만해지고(예: softmax([5.00,6.25,5.625]) ≈ [0.157,0.549,0.294]) 학습이 빠르고 안정적으로 진행된다.

### E10. Multi-Head Attention이 필요한 이유와, head 수를 무한히 늘릴 수 없는 이유를 함께 설명하시오.

> 필요한 이유: 같은 단어라도 문법적 관계·의미적 맥락·시제 등 여러 이유로 다른 단어에 주목해야 하는데, 단일 head는 한 가지 관점의 관계만 포착한다. 여러 head를 두면 다양한 관점을 동시에 볼 수 있다.
>
> 늘릴 수 없는 이유: 총 차원 d를 h로 나눠 쓰므로 head당 차원이 d/h로 반비례해 줄어든다(d=512, h=64면 8차원). 표현이 거칠어질 뿐 아니라, 일정 수준을 넘으면 서로 비슷한 패턴만 학습하는 중복 헤드가 늘어난다. 따라서 데이터 규모·태스크에 따라 최적 head 수가 따로 존재한다.

### E11. Encoder / Encoder-Decoder / Decoder 세 구조를 사전학습 방법과 함께 비교하시오.

> Encoder(BERT): 양방향 문맥 활용. Masked LM(15%, 80/10/10) + NSP로 사전학습. 이해 태스크(분류·NER·QA)에 강하고 생성에는 부적합.
>
> Encoder-Decoder(T5): 인코더가 입력 전체를 보고 디코더가 생성. Span Corruption으로 사전학습. 모든 태스크를 Text-to-Text로 통일해 이해와 생성을 모두 커버.
>
> Decoder(GPT): 단방향(masked) 문맥. Next Token Prediction으로 사전학습. 생성에 자연스럽게 적합하며, 사전학습 목표와 사용 방식이 동일해 파인튜닝 없이도 쓸 수 있다.

### E12. BERT의 MLM에서 선택된 토큰을 100% [MASK]로 바꾸지 않고 80/10/10으로 나누는 이유를 설명하시오.

> [MASK] 토큰은 사전학습에만 등장하고 파인튜닝·추론 시에는 나타나지 않는다. 만약 100% [MASK]로 하면 모델은 '[MASK]가 보일 때만 문맥을 참고한다'고 학습해, 실제 사용 시 그 능력이 발휘되지 않는다(pretrain–finetune discrepancy). 그래서 10%는 엉뚱한 단어로 바꿔 눈에 보이는 단어도 못 믿게 만들고, 10%는 그대로 두되 예측은 시켜 [MASK] 없이도 좋은 표현을 만들게 한다. 결과적으로 모델은 모든 위치에서 항상 문맥을 반영한 표현을 만들게 된다.

### E13. T5의 Span Corruption이 BERT의 단일 토큰 마스킹과 다른 점과, 그 차이가 만드는 능력을 설명하시오.

> BERT는 토큰을 하나씩 독립적으로 가리지만 T5는 연속된 구간(span)을 통째로 지우고 <X>, <Y> 같은 placeholder로 치환한다. 차이는 두 가지다.
>
> ① 한 토큰만 가리면 주변으로 쉽게 유추되어 학습 신호가 약하지만, 구간을 지우면 더 넓은 문맥과 구 단위 의미를 동원해야 복원할 수 있다.
>
> ② 복원해야 할 정답의 길이가 정해져 있지 않으므로, 모델은 '무엇을 쓸지'뿐 아니라 '언제 멈출지'까지 배워야 한다. 이것이 곧 문장 생성 능력이며, BERT가 갖지 못한 부분이다.

### E14. In-Context Learning이 기존의 파인튜닝과 근본적으로 무엇이 다른지 설명하시오.

> 파인튜닝은 태스크별 레이블 데이터로 gradient를 흘려 모델 가중치를 실제로 갱신하고, 태스크마다 별도의 모델 사본이 필요하다. In-Context Learning은 gradient update가 전혀 없다. 입력 텍스트(컨텍스트)에 태스크 설명과 예시를 넣어주면 모델이 그 패턴을 따라 답한다. 따라서 ① 학습 데이터·시간·모델 사본이 필요 없고 ② 하나의 모델로 여러 태스크를 즉시 수행할 수 있다. 다만 이 능력은 모델 크기가 커야 뚜렷하게 나타난다(1.3B에서는 거의 없고 175B에서 강하게 나타남).

### E15. Chain-of-Thought prompting이 왜 효과가 있는지, 그리고 Zero-shot CoT가 시사하는 바를 설명하시오.

> 단순 few-shot은 '문제 → 정답'만 보여주므로 모델도 답을 바로 내려 하고, 여러 단계 계산이 필요한 문제에서 틀린다. CoT는 예시의 답에 풀이 과정을 넣어 모델도 중간 단계를 생성하게 만든다. 중간 단계를 텍스트로 출력하면 그것이 다음 단계의 입력이 되므로, 사실상 계산을 여러 스텝에 나눠 수행하는 효과가 생긴다(GSM8K 18% → 57%).
>
> Zero-shot CoT는 예시 없이 "Let's think step by step" 한 문장만 붙여 17.7 → 78.7을 달성했다. 가중치가 하나도 바뀌지 않았으므로, 이 추론 능력은 원래부터 모델 안에 있었고 단지 꺼내지지 않고 있었다는 뜻이다. 즉 프롬프팅은 능력을 '가르치는' 것이 아니라 '끌어내는(elicit)' 작업에 가깝다.
