### TODO 4: IQR 기반 이상치 탐지 함수- **요구사항**: 주어진 컬럼에서 IQR 기준 이상치 **행들**을 반환하는 함수를 완성하세요.

```python
def detect_outliers_iqr(data, column, k=1.5):    Q1 = data[column].quantile(0.25)    Q3 = data[column].quantile(0.75)    IQR = Q3 - Q1    lower_bound = Q1 - k * IQR    upper_bound = Q3 + k * IQR    return data[(data[column] < lower_bound) | (data[column] > upper_bound)]test = pd.DataFrame({"v": [3, 7, 8, 5, 12, 14, 21, 13, 18, 200]})res = detect_outliers_iqr(test, "v")assert len(res) == 1 and int(res["v"].iloc[0]) == 200, f"이상치 1개(200)여야 합니다. got {res}"assert len(detect_outliers_iqr(test, "v", k=100)) == 0, "k를 키우면 이상치가 없어야 합니다."print("✅ TODO 4 통과")
```

### TODO 9: 평가지표를 직접 계산하기혼동 행렬로부터 accuracy / precision / recall / f1 을 **직접** 계산하는 함수를 완성하세요.(sklearn 함수를 쓰지 말고 공식으로!)

```python
def metrics_from_cm(cm):    """cm = [[TN, FP], [FN, TP]] → (accuracy, precision, recall, f1)"""    TN, FP = cm[0][0], cm[0][1]    FN, TP = cm[1][0], cm[1][1]    accuracy  = (TP + TN) / (TP + TN + FP + FN)    precision = TP / (TP + FP) if (TP + FP) else 0.0    recall    = TP / (TP + FN) if (TP + FN) else 0.0    f1 = 2*precision*recall/(precision+recall) if (precision+recall) else 0.0    return accuracy, precision, recall, f1a, p, r, f = metrics_from_cm([[80, 20], [10, 90]])assert abs(a-0.85) < 1e-9,   f"accuracy: {a}"assert abs(p-90/110) < 1e-9, f"precision: {p}"assert abs(r-0.90) < 1e-9,   f"recall: {r}"assert abs(f-2*(90/110)*0.9/((90/110)+0.9)) < 1e-9, f"f1: {f}"print(f"✅ TODO 9 통과  acc={a:.4f} prec={p:.4f} rec={r:.4f} f1={f:.4f}")
```

### TODO 12: 회귀 평가지표 직접 구현RMSE, MAE, $R^2$ 를 numpy 로 직접 계산하세요. (sklearn 함수 금지)$$\text{RMSE}=\sqrt{\tfrac1n\sum(y-\hat y)^2},\quad\text{MAE}=\tfrac1n\sum|y-\hat y|,\quadR^2=1-\frac{\sum(y-\hat y)^2}{\sum(y-\bar y)^2}$$

```python
def regression_metrics(y_true, y_pred):    y_true = np.asarray(y_true, dtype=float)    y_pred = np.asarray(y_pred, dtype=float)    err = y_true - y_pred    rmse = np.sqrt((err ** 2).mean())    mae  = np.abs(err).mean()    r2   = 1 - (err ** 2).sum() / ((y_true - y_true.mean()) ** 2).sum()    return rmse, mae, r2yt = np.array([10., 20, 30, 40, 50])yp = np.array([12., 18, 33, 39, 47])rmse, mae, r2 = regression_metrics(yt, yp)from sklearn.metrics import mean_absolute_error, r2_scoreassert abs(rmse - np.sqrt(((yt-yp)**2).mean())) < 1e-9, f"rmse: {rmse}"assert abs(mae - mean_absolute_error(yt, yp)) < 1e-9,   f"mae: {mae}"assert abs(r2 - r2_score(yt, yp)) < 1e-9,               f"r2: {r2}"print(f"✅ TODO 12 통과  RMSE={rmse:.4f}  MAE={mae:.4f}  R2={r2:.4f}")
```

### 파트 F 모범답안**F1.** 모델은 주어진 데이터의 패턴을 그대로 학습하므로, 데이터가 잘못돼 있으면 아무리 좋은 모델을 써도잘못된 결과가 나온다(“Trash in, trash out”). EDA는 샘플 수·타입·결측·이상치·분포·상관을 먼저 확인해전처리 방향과 문제 정의를 결정하는 단계다. 실무에서 모델링보다 데이터 정제에 시간이 더 드는 이유이기도 하다.**F2.** 상관은 두 변수가 함께 움직인다는 관찰일 뿐, 한쪽이 다른 쪽의 원인임을 보장하지 않는다.여름에 아이스크림 판매량과 익사 사고가 함께 증가하지만, 원인은 둘 다에 영향을 주는 **기온**(숨은 변수)이다.또한 역인과(방향이 반대)나 우연한 상관도 가능하다. 인과를 주장하려면 실험(무작위 배정)이나 인과추론 설계가 필요하다.**F3.** ① 평균·표준편차는 이상치 자체에 크게 끌려가지만, $Q_1,Q_3$는 순위 기반이라 **극단값에 덜 민감**하다.② 정규분포를 가정하지 않으므로 **치우친(skewed) 분포에도 적용**할 수 있다.**F4.** 데이터 누수는 “평가 시점에는 알 수 없는 정보”가 학습 과정에 새어 들어가는 것이다.전체 데이터로 스케일링·결측 대체·특성 선택을 하면 test의 통계가 train에 반영되어,테스트 성능이 실제보다 낙관적으로 나온다. 따라서 **분할 → train으로 fit → test는 transform** 순서를 지켜야 하고,`Pipeline`으로 묶으면 교차검증 안에서도 자동으로 지켜진다.**F5.** 9:1 불균형에서는 전부 다수 클래스로 찍어도 accuracy가 0.9다.소수 클래스를 얼마나 잘 잡는지 보려면 **Precision / Recall / F1**을,임계값과 무관한 전체 판별력을 보려면 **ROC-AUC**(또는 불균형이 매우 심하면 **PR-AUC**)를 봐야 한다.비용이 비대칭이면(예: 암 진단) Recall을 우선하는 식으로 목표 지표를 먼저 정하는 것이 중요하다.**F6.** 큰 오차를 특히 피해야 하는 문제(예: 재고 부족, 안전 관련)라면 큰 오차를 제곱해 크게 벌하는 **RMSE**,이상치가 많고 “평균적인 오차 크기”가 궁금하면 **MAE**가 적절하다. 둘의 비율(RMSE/MAE)이 크면큰 오차가 몇 개 섞여 있다는 신호다. $R^2$는 단위가 없어 **평균 예측 대비 얼마나 개선했는지**를 보여주는상대 지표이며, 서로 다른 스케일의 문제를 비교하거나 모델의 설명력을 요약할 때 쓴다.
