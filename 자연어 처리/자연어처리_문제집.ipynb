## 파트 A. O / X (26문항)

각 문장이 맞으면 `'O'`, 틀리면 `'X'` 를 답안에 적으세요.

**A1.** 서로 다른 두 단어의 원-핫 벡터는 내적이 항상 0이다.

**A2.** 워드 임베딩의 차원 수는 반드시 어휘 크기와 같아야 한다.

**A3.** CBOW는 중심 단어로 주변 단어를 예측한다.

**A4.** Skip-gram은 CBOW보다 학습 속도가 빠르다.

**A5.** RNN은 모든 시점에서 서로 다른 가중치 행렬을 사용한다.

**A6.** RNN의 기울기 소실은 역전파에서 1보다 작은 값이 반복해서 곱해지기 때문에 발생한다.

**A7.** LSTM에서 cell state는 short-term 정보를, hidden state는 long-term 정보를 저장한다.

**A8.** LSTM의 forget gate 값이 1에 가까우면 이전 cell state가 거의 그대로 유지된다.

**A9.** 4-gram 언어모델은 문장의 맨 앞 단어까지 모두 조건으로 사용한다.

**A10.** N-gram 언어모델에서 n을 크게 하면 항상 성능이 좋아진다.

**A11.** Seq2Seq에서 인코더와 디코더는 서로 독립적으로 따로 학습된다.

**A12.** Teacher Forcing은 디코더 입력으로 모델의 예측 대신 정답 단어를 넣는 기법이다.

**A13.** Greedy Inference는 한 번 잘못 고른 단어를 나중에 되돌릴 수 있다.

**A14.** Attention은 Seq2Seq의 bottleneck problem을 완화한다.

**A15.** Self-Attention에서는 Query, Key, Value가 모두 같은 문장에서 만들어진다.

**A16.** Self-Attention은 단어의 순서 정보를 자체적으로 반영한다.

**A17.** Masked Self-Attention은 미래 단어의 attention score를 -∞로 설정한다.

**A18.** Scaled Dot Product에서 √(d/h)로 나누는 이유는 계산 속도를 높이기 위해서다.

**A19.** Attention Head 수를 늘리면 성능이 무한히 향상된다.

**A20.** Transformer의 encoder는 양방향 문맥을, decoder는 단방향 문맥을 사용한다.

**A21.** BERT는 입력 토큰의 15%를 선택해 그 중 80%를 [MASK]로 바꾼다.

**A22.** BERT는 텍스트 생성 태스크에 가장 적합한 모델이다.

**A23.** T5는 모든 태스크를 Text-to-Text 형식으로 통일했다.
