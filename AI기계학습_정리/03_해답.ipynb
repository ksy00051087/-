### D6. 로지스틱 회귀 학습 (경사하강)음의 로그우도의 기울기는 $\frac{1}{n}\mathbf X^\top(\mathbf p-\mathbf y)$ 입니다.

```python
def fit_logistic(X, y, lr=0.5, iters=20000):    beta = np.zeros(X.shape[1])    for _ in range(iters):        p = sigmoid(X @ beta)        grad = X.T @ (p - y) / len(y)        beta -= lr*grad    return betarr = np.random.default_rng(2)xb = rr.uniform(-3, 3, 600)yb = (rr.random(600) < sigmoid(0.5 + 2.0*xb)).astype(float)Xb = np.column_stack([np.ones(600), xb])bb = fit_logistic(Xb, yb)assert np.allclose(bb, [0.5, 2.0], atol=0.35), f"추정 틀림: {bb}"print("D6 통과 ✔", np.round(bb,3), " (참값 [0.5, 2.0])")
```

### D8. Deep 네트워크 순전파 (행렬 표기)$$\mathbf h_{k+1}=\mathrm a[\boldsymbol\beta_k+\boldsymbol\Omega_k\mathbf h_k],\qquad \text{마지막 층은 활성화 없음}$$

```python
def gd_step(x, y, phi0, phi1, alpha):    e = phi0 + phi1*x - y    g0 = 2*e.mean()    g1 = 2*(x*e).mean()    return phi0 - alpha*g0, phi1 - alpha*g1xg = np.array([1.,2,3]); yg = np.array([2.,3,5])e = 0.0+1.0*xg-ygg0 = 2*e.mean(); g1 = 2*(xg*e).mean()assert np.allclose(gd_step(xg,yg,0.0,1.0,0.1), (0.0-0.1*g0, 1.0-0.1*g1))print("D9 통과 ✔", np.round(gd_step(xg,yg,0.0,1.0,0.1),4))
```

### D10. 역전파 (2층 네트워크, 수치미분으로 자동 검증)$$x\to f_0=\beta_0+\omega_0x\to h_1=\mathrm{ReLU}[f_0]\to f_1=\beta_1+\omega_1h_1\to \ell=(f_1-y)^2$$$\partial\ell/\partial\omega_0,\ \partial\ell/\partial\beta_0,\ \partial\ell/\partial\omega_1,\ \partial\ell/\partial\beta_1$ 을 구하세요.

```python
def backprop(x, y, w0, b0_, w1, b1_):    # 순전파    f0 = b0_ + w0*x    h1 = max(0.0, f0)    f1 = b1_ + w1*h1    # 역전파 (연쇄법칙)    dl_df1 = 2*(f1 - y)    dw1 = dl_df1 * h1    db1 = dl_df1    dl_dh1 = w1 * dl_df1    dl_df0 = dl_dh1 * (1.0 if f0 > 0 else 0.0)   # ReLU 미분    dw0 = dl_df0 * x    db0 = dl_df0    return dw0, db0, dw1, db1def loss_fn(x,y,w0,b0_,w1,b1_):    return (b1_ + w1*max(0.0, b0_+w0*x) - y)**2def check(args):    g = backprop(*args); eps=1e-6; num=[]    for i in range(2,6):        a=list(args); a[i]+=eps; up=loss_fn(*a)        a=list(args); a[i]-=eps; dn=loss_fn(*a)        num.append((up-dn)/(2*eps))    assert np.allclose(g, num, atol=1e-5), "역전파와 수치미분이 다릅니다"    return np.round(g,6)print("case1 (f0>0)          :", check((1.0, 3.0, 1.5, -0.5, 2.0, 0.0)))print("case2 (f0<0, ReLU 죽음):", check((1.0, 3.0, -1.5, -0.5, 2.0, 0.0)))print("D10 통과 [OK]  역전파가 수치미분과 일치합니다")
```
